- Циклом в конце перебираем значения

```python
def f(N):
    N3 = to_sys(N, 3)

    if N % 3 == 0:
        N3 = "1" + N3 + "02"
    else:
        r3 = to_sys((N % 3) * 4, 3)
        N3 = N3 + r3

    R = int(N3, 3)
    return {"N": N, "N3": N3, "R": R}


# for i in range(100):
#     if f(i)["R"] <= 250:
#         print(i)
```

- Агоритм перевода в **ИЗ ДЕСЯТИЧКИ В ЛЮБУЮ СИСТЕМУ СЧИСЛЕНИЯ**, не представленную в дефолтных функциях питона, в остальных случаях использовать встроенные функции типо `bin`, `oct`, `hex`

```python
def to_sys(N, base):
    if N == 0:
        return "0"
    
    sys_N = ""
    digits = "0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ"

    while N > 0:
        remainder = N % base
        
        sys_N = digits[remainder] + sys_N
        N //= base
    
    return sys_N

to_sys(27284841127455234619139841725, 25)
```

### **Задание №2**

- Перебор всех значений переменных, потом сравнить аналитически с таблицей, искать противоречия

```python
print("w z y x | F")

for x in range(2):
    for y in range(2):
        for z in range(2):
            for w in range(2):
                F = (x == (not(y))) <= ((x and w) == (z and not(w)))
                if F == 0 and z == 1:

                    print(f"{w} {z} {y} {x} | {int(F)}")
```

# УНИВЕРСАЛЬНЫЙ АЛГОРИТМ В ДЕСЯТИЧКУ + РЕШЕНИЕ К ЗАДАЧЕ

```python
def to10(a):
    res = 0
    s22 = 1

    for i in range(len(a) - 1, -1, -1):
        res += a[i] * s22
        s22 *= 22

    return res

for x in range(22):
    a = to10([2, 7, x, 9, 8, 8, 7, 6])
    b = to10([2, 6, x, 5, 1])
    c = to10([7, 1, 1, x, 5])

    f = a + b + c

    if f % 21 == 0:
        print(x, f / 21)
```
