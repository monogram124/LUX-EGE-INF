### **Задание №5**

- Создать функцию, туда передавать число N, на выходе возвращать словарик с локальными перменными, значение которых интересно посмотреть
- Помнить, что теперь результат - это словарик, чтобы получить R пишем `f(i)["R"]`
- Циклом в конце перебираем значения

```python
def f(N):
    N3 = to_sys(N, 3)

    if N % 3 == 0:
        N3 = "1" + N3 + "02"
    else:
        r3 = to_sys((N % 3) * 4, 3)
        N3 = N3 + r3

    R = int(N3, 3)
    return {"N": N, "N3": N3, "R": R}


for i in range(100):
    if f(i)["R"] <= 250:
        print(i)
```

- Агоритм перевода в **ИЗ ДЕСЯТИЧКИ В ЛЮБУЮ СИСТЕМУ СЧИСЛЕНИЯ**, не представленную в дефолтных функциях питона, в остальных случаях использовать встроенные функции типо `bin`, `oct`, `hex`

```python
def to_sys(N, base):
    if N == 0:
        return "0"
    
    sys_N = ""
    digits = "0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ"

    while N > 0:
        remainder = N % base
        
        sys_N = digits[remainder] + sys_N
        N //= base
    
    return sys_N
```
